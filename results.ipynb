```python
import matplotlib.pyplot as plt

# Define the validation accuracies per epoch for the CNN model,
# extracted from the logs:

# For 10% of the dataset:
val_acc_10 = [0.7494, 0.7312, 0.7831, 0.7613, 0.7706, 0.7663, 0.6963, 0.7837, 0.7987, 0.7844]

# For 20% of the dataset:
val_acc_20 = [0.6556, 0.7844, 0.7862, 0.7706, 0.7825, 0.7734, 0.8091, 0.8022, 0.8091, 0.8059]

# For 30% of the dataset:
val_acc_30 = [0.7919, 0.7975, 0.8067, 0.8333, 0.8421, 0.8465, 0.8425, 0.8558, 0.8494, 0.8440]

# Define the epochs (since you trained for 10 epochs):
epochs = list(range(1, 11))

# Create the plot:
plt.figure(figsize=(10, 6))
plt.plot(epochs, val_acc_10, marker='o', label='CNN with 10% Data')
plt.plot(epochs, val_acc_20, marker='o', label='CNN with 20% Data')
plt.plot(epochs, val_acc_30, marker='o', label='CNN with 30% Data')

plt.title('CNN Convergence: Validation Accuracy vs. Epochs')
plt.xlabel('Epoch')
plt.ylabel('Validation Accuracy')
plt.xticks(epochs)
plt.ylim(0.6, 0.9)
plt.legend()
plt.grid(True)
plt.tight_layout()

# Save the plot as an image file:
plt.savefig("cnn_convergence_comparison.png")
plt.show()
```
